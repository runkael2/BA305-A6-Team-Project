```python
# Libraries used
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.metrics import accuracy_score, confusion_matrix, mean_squared_error, r2_score
```

```python
url = "https://raw.githubusercontent.com/runkael2/BA305-A6-Team-Project/refs/heads/main/Rachael/UK_Gender_Pay_Gap_workbook_allyears.csv"
paygap_df = pd.read_csv(url)

paygap_df = paygap_df.drop(columns=['Address', 'PostCode','CompanyNumber','ResponsiblePerson',
                                    'DueDate','DateSubmitted','CompanyLinkToGPGInfo',
                                    'PostCodeAbbrv', 'CurrentName'])
# getting rid of most irrelevant columns for this prediction
display(paygap_df.head(5))

```

**Data Cleaning**



```python
# Missing Regions get marked as "unknown"
# approximately 200 data points with unknown regions

paygap_df["REGION"] = (
    paygap_df["REGION"]
    .astype("string")
    .str.strip()
    .replace("NOT FOUND", "Unknown")
    .fillna("Unknown")
)

# display(paygap_df.head(200))
```

```python
before = len(paygap_df)

model_df = paygap_df.dropna(
    subset=["DiffMedianHourlyPercent"]
).copy()

removed = before - len(model_df)

print("Rows removed:", removed)
print("Rows remaining:", len(model_df))

# No missing target values (nice)
```
